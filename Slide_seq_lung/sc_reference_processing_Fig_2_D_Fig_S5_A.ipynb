## Load dataset

The data was obtained from https://singlecell.broadinstitute.org/single_cell/study/SCP1288/tumor-and-immune-reprogramming-during-immunotherapy-in-advanced-renal-cell-carcinoma#/.

```python
adata = sc.read_mtx(os.path.join(SC_REF_PATH, 'SCP1288/expression/60c76a18771a5b0ba10ea91b/matrix.mtx.gz')).T
barcodes = pd.read_csv(os.path.join(SC_REF_PATH, 'SCP1288/expression/60c76a18771a5b0ba10ea91b/barcodes.tsv.gz'), sep="\t", compression="gzip", index_col=0, header=None).index.tolist()
features = pd.read_csv(os.path.join(SC_REF_PATH, 'SCP1288/expression/60c76a18771a5b0ba10ea91b/features.tsv.gz'), sep="\t", compression="gzip", index_col=0, header=None).index.tolist()

adata.obs_names = barcodes
adata.var_names = features
```

```python
adata
```

```python
metadata_df = pd.read_csv(os.path.join(SC_REF_PATH, 'SCP1288/metadata/Final_SCP_Metadata.txt'), sep="\t", index_col=0)
cluster_df = pd.read_csv(os.path.join(SC_REF_PATH, 'SCP1288/cluster/Final_SCP_ClusterFile.txt'), sep="\t", index_col=0)
```

```python
cluster_df = cluster_df.drop('TYPE', axis=0)
metadata_df = metadata_df.drop('TYPE', axis=0)
```

```python
adata.obs[metadata_df.columns] = metadata_df
adata = adata[adata.obs.dropna().index].copy()
```

```python
adata.layers['counts'] = adata.X.copy()
```

## Visualize macrophage subset expression

```python
all_adata = sc.read_h5ad(os.path.join(ADATA_PATH, 'Slide_seq_lung_sc_ref_scVI_adata.h5ad'))
```

```python
macrophage_subsets = ['CXCL10-Hi TAM', 'VSIR-Hi TAM', 'GPNMB-Hi TAM', 'FOLR2-Hi TAM', 'Cycling TAM']
```

```python
lung_adata = sc.read_h5ad(os.path.join(ADATA_PATH, 'Slide_seq_lung_sc_ref_revision_scVI_adata.h5ad'))
```

```python
mac_adata = all_adata[all_adata.obs['FinalCellType'].isin(macrophage_subsets)].copy()
```

```python
norm = sc.pp.normalize_total(mac_adata, target_sum=1e4, inplace=False)
mac_adata.layers["normalized"] = norm["X"]
mac_adata.layers["log_norm"] = sc.pp.log1p(norm["X"], copy=True)
```

```python
macrophage_markers = ['CXCL9', 'CXCL10', 'CXCL11', 'GBP1', 'IRF1', 'HLA-DPB1', 'CD86',
                      'CXCR4', 'VSIR', 'HLA-DPA1',
                      'GPNMB', 'MARCO', 'CTSO', 'CTSB',
                      'SELENOP', 'FOLR2', 'IL2RA', 'F13A1',
                      'PCLAF', 'STMN1', 'TUBB',
                      'SLC8A1']
```

```python
print('\n'.join(macrophage_markers))
```

```python
mac_adata.obs["FinalCellType"] = mac_adata.obs["FinalCellType"].cat.reorder_categories(macrophage_subsets)
```

```python
sc.pl.dotplot(mac_adata, macrophage_markers, "FinalCellType", dendrogram=False, standard_scale='var', layer='log_norm', cmap='YlGnBu', swap_axes=False, show=False)
plt.savefig(os.path.join(PLOTS_PATH, "Macrophage_subset_expr_all_dot_plot.pdf"))
```
